**2. Differentiations: Differentiate between the following: ○ Mutable and immutable data types ○ List and tuple**

ANS:

Mutable vs Immutable Data Types
Mutable data types are those whose values can be changed after creation. For example, a list can be modified by adding or removing elements. Immutable data types cannot be changed once they are created.

* Mutable: Can be changed (e.g., List, Dictionary)

* Immutable: Cannot be changed (e.g., Tuple, String)

**List vs Tuple**

Lists and tuples are used to store multiple values, but they are different in nature.

->List: Mutable, uses square brackets [ ]

->Tuple: Immutable, uses parentheses ( )

Lists are flexible, while tuples are faster and fixed

**17.Slicing Practice: Given a list of numbers, write a program that uses slicing to create a new list containing alternate elements. Print both the original and sliced list.**

ANS:
*Step 1:* Understand the question

* “alternate elements” means:

* Take one element
* Skip one
* Take next

Example:
[1, 2, 3, 4, 5, 6] → [1, 3, 5]

*Step 2:* Think slicing

Slicing format:


```
list[start : end : step]
```
Now the code is:


```
# Original list
nums = [1, 2, 3, 4, 5, 6]

# Slicing to get alternate elements
alt = nums[::2]

# Print both lists
print("Original list:", nums)
print("Alternate elements:", alt)
```
OUTPUT:

Original list: [1, 2, 3, 4, 5, 6]

Alternate elements: [1, 3, 5]

**1. Loop Logic: Why do programming languages provide both for and while loops instead of only one looping structure?**

ANS:
* Programming languages provide both for and while loops because they are used in different situations.

* A for loop is used when we already know how many times we want to repeat something. A while loop is used when we don't know the number of repetitions and the loop depends on a condition.

* for loop → fixed number of times
* while loop → depends on condition

*Example*:


```
for i in range(5):  # fixed
    print(i)

x = 0
while x < 5:   # condition based
    print(x)
    x += 1
```
