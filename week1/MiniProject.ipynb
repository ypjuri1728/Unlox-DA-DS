#**Project 1: Console-Based Smart Calculator**

*Problem Statement Create a menu-driven calculator that performs multiple mathematical operations using Python and NumPy*

###Logic Explanation:

* Display a menu of operations
* Take user choice
* Ask for number of inputs
* Store numbers in list and convert to NumPy array
* Perform selected operation
* Display result
* Repeat until user exits

```python
import numpy as np

def calculator():
    while True:
        print("\n--- Smart Calculator ---")
        print("1. Addition")
        print("2. Subtraction")
        print("3. Multiplication")
        print("4. Division")
        print("5. Power")
        print("6. Modulus")
        print("7. Exit")

        choice = int(input("Enter your choice: "))

        if choice == 7:
            print("Exiting calculator...")
            break

        # Input numbers
        n = int(input("How many numbers? "))
        nums = []

        for i in range(n):
            num = float(input(f"Enter number {i+1}: "))
            nums.append(num)

        arr = np.array(nums)

        # Operations
        if choice == 1:
            print("Result:", np.sum(arr))

        elif choice == 2:
            result = arr[0]
            for i in arr[1:]:
                result -= i
            print("Result:", result)

        elif choice == 3:
            result = 1
            for i in arr:
                result *= i
            print("Result:", result)

        elif choice == 4:
            result = arr[0]
            try:
                for i in arr[1:]:
                    result /= i
                print("Result:", result)
            except ZeroDivisionError:
                print("Error: Division by zero")

        elif choice == 5:
            if len(arr) >= 2:
                print("Result:", arr[0] ** arr[1])
            else:
                print("Need at least 2 numbers")

        elif choice == 6:
            if len(arr) >= 2:
                print("Result:", arr[0] % arr[1])
            else:
                print("Need at least 2 numbers")

        else:
            print("Invalid choice")

# Run program
calculator()
```

#**Project 2: Student Marks Analyzer**

*Problem Statement Develop a program that analyzes student marks using Python and NumPy.*

This project is used to analyze marks of students using Python and NumPy. The program takes marks as input from the user, stores them in a NumPy array, and calculates total marks, average marks, highest and lowest marks. It also assigns grades based on the average marks using conditional statements.

**Logic Explanation**
* Take number of students
* Take marks input using loop
* Store marks in list
* Convert list to NumPy array

**Calculate:**
* Total marks
* Average marks
* Highest marks
* Lowest marks
* Assign grade using if-else
* Display results

```python
import numpy as np

# Function to assign grade
def get_grade(avg):
    if avg > 85:
        return "A"
    elif avg >= 70:
        return "B"
    elif avg >= 50:
        return "C"
    else:
        return "Fail"

# Main program
n = int(input("Enter number of students: "))

marks_list = []

# Input marks
for i in range(n):
    marks = float(input(f"Enter marks of student {i+1}: "))
    marks_list.append(marks)

# Convert to NumPy array
marks_array = np.array(marks_list)

# Calculations
total = np.sum(marks_array)
average = np.mean(marks_array)
highest = np.max(marks_array)
lowest = np.min(marks_array)

# Grade calculation
grade = get_grade(average)

# Output
print("\n--- Result ---")
print("Total Marks:", total)
print("Average Marks:", average)
print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
```

#Project 3: Daily Expense Tracker
*Problem Statement Create a program to track daily expenses and analyze spending.*

**Logic Explanation**

* Take number of days
* Take expense input for each day
* Store values in list
* Convert list into NumPy array

**Calculate:**
* Total expense
* Average expense
* Highest expense
* Compare average with limit
* Display summary

```python
import numpy as np

# Main program
n = int(input("Enter number of days: "))

expense_list = []

# Input expenses
for i in range(n):
    expense = float(input(f"Enter expense for day {i+1}: "))
    expense_list.append(expense)

# Convert to NumPy array
expense_array = np.array(expense_list)

# Calculations
total = np.sum(expense_array)
average = np.mean(expense_array)
highest = np.max(expense_array)

# Set limit
limit = 500

# Output
print("\n--- Expense Summary ---")
print("Total Expense:", total)
print("Average Expense:", average)
print("Highest Expense:", highest)

# Warning condition
if average > limit:
    print("Warning: Your average expense is high!")
else:
    print("Your expenses are under control.")
```
