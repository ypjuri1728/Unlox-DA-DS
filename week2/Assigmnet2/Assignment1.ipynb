**1. Data Loading and Inspection**

● Load the dataset into a Pandas DataFrame.

● Display the first 10 rows and inspect the basic structure.

● Report the number of rows and columns.

● Identify data types and missing values

```python
import pandas as pd

# Load dataset
df = pd.read_csv('/mnt/data/SuperMarket Analysis.csv')

# First 10 rows
print(df.head(10))

# Shape (rows, columns)
print("Shape:", df.shape)

# Data types and info
print(df.info())

# Missing values
print(df.isnull().sum())
```

Explanation:

* head(10) → first 10 rows
* shape → total rows & columns
* info() → data types
* isnull() → missing values check

**2. Data Cleaning and Preparation**

● Handle any missing values appropriately.

● Convert the Date column to a datetime type.

● Create a new column TotalSales which represents the overall purchase amount if not already present

```python
# Handle missing values
df = df.dropna()   # or use fillna() if needed

# Convert Date column
df['Date'] = pd.to_datetime(df['Date'])

# Create TotalSales column (if not present)
if 'TotalSales' not in df.columns:
    df['TotalSales'] = df['Quantity'] * df['Unit price']
```

Explanation:

* Missing values removed using dropna()
* Date converted to datetime → important for time analysis
* TotalSales = Quantity × Unit Price

**3. Exploratory Data Analysis**

● Identify the top 3 product lines contributing the most to total sales.

● Determine which branch has the highest average sales.

● Find the distribution of customer types (Member vs Normal) and identify which group spends more on average.

Top 3 Product Lines

```python
top_products = df.groupby('Product line')['TotalSales'].sum().sort_values(ascending=False).head(3)
print(top_products)
```

Branch with Highest Average Sales

```python
branch_avg = df.groupby('Branch')['TotalSales'].mean().sort_values(ascending=False)
print(branch_avg)
```

Customer Type Distribution + Spending

```python
# Count
print(df['Customer type'].value_counts())

# Average spending
customer_spending = df.groupby('Customer type')['TotalSales'].mean()
print(customer_spending)
```

**4. Visualizations**

● Plot a bar chart showing total sales by product line.

● Create a box plot to compare sales distribution across branches.

● Draw a histogram for the TotalSales column.

● Use Seaborn pairplot for numerical attributes to identify relationships.

● Plot a correlation heatmap for all numerical features.

mport libraries

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

Bar Chart: Sales by Product Line

```python
df.groupby('Product line')['TotalSales'].sum().plot(kind='bar')
plt.title("Total Sales by Product Line")
plt.show()
```

Box Plot: Sales by Branch

```python
sns.boxplot(x='Branch', y='TotalSales', data=df)
plt.title("Sales Distribution Across Branches")
plt.show()
```

Histogram: Total Sales

```python
plt.hist(df['TotalSales'])
plt.title("Distribution of Total Sales")
plt.show()
```

Pairplot

```python
sns.pairplot(df.select_dtypes(include=['float64','int64']))
plt.show()
```

Correlation Heatmap

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

1. Consistent Product Line

```python
consistency = df.groupby(['Product line','Branch'])['TotalSales'].mean().unstack()
print(consistency)
```

Product line with similar values across branches = consistent

2. Gender Influence

```python
gender_sales = df.groupby('Gender')['TotalSales'].mean()
print(gender_sales)
```

Compare averages:

* If difference is small → no strong influence
* If large → gender affects spending
