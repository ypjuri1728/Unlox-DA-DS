## 🎯 Problem Statement
Individuals often track expenses but struggle to understand spending patterns and control unnecessary costs. This project focuses on analyzing personal expense data to identify trends and spending behavior.

---

## 🎯 Objectives
- Analyze monthly expense data  
- Identify high-spending categories  
- Understand daily and category-wise expense patterns  
- Visualize spending behavior  

##Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

##Load Dataset

### 📂 Step 1: Load Dataset

```python
df = pd.read_csv("SuperMarket Analysis.csv")
df.head(10)
```

##Data Inspection
##### 🔍 Step 2: Inspect Data Structure

```python
df.info()
df.describe()
df.isnull().sum()
```

##Data Cleaning

### 🧹 Step 3: Data Cleaning and Preprocessing

```python
# Fill missing values
df.fillna(method='ffill', inplace=True)

# Convert Date column
df['Date'] = pd.to_datetime(df['Date'])

# Create new columns
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['Day_Name'] = df['Date'].dt.day_name()

df.head()
```

### 💰 Step 4: Total and Average Expense Analysis

```python
total_expense = df['Total'].sum()
avg_expense = df['Total'].mean()

print("Total Expense:", total_expense)
print("Average Expense:", avg_expense)
```

###  📊 Step 5: Category-wise Expense Analysis

```python
category_expense = df.groupby('Product line')['Total'].sum().sort_values(ascending=False)
category_expense
```

### 📈 Step 6: Monthly Expense Trend

```python
monthly_expense = df.groupby('Month')['Total'].sum()

plt.figure()
monthly_expense.plot(marker='o')
plt.title("Monthly Expense Trend")
plt.xlabel("Month")
plt.ylabel("Total Expense")
plt.show()
```

### 📅 Step 7: Daily Spending Pattern

```python
daily_expense = df.groupby('Day_Name')['Total'].sum()

plt.figure()
daily_expense.plot(kind='bar')
plt.title("Daily Spending Pattern")
plt.xlabel("Day")
plt.ylabel("Total Expense")
plt.show()
```

### 🥧 Step 8: Category-wise Visualization

####Pie chat:

```python
plt.figure()
category_expense.plot(kind='pie', autopct='%1.1f%%')
plt.title("Category-wise Spending")
plt.ylabel("")
plt.show()
```

####Bar Chart

```python
plt.figure()
category_expense.plot(kind='bar')
plt.title("Category-wise Spending")
plt.xlabel("Category")
plt.ylabel("Total Expense")
plt.show()
```

### 📦 Step 9: Outlier Detection

```python
plt.figure()
sns.boxplot(x=df['Total'])
plt.title("Expense Outliers")
plt.show()
```

### 🔍 Step 10: Insights and Findings

```python
print("Highest Spending Category:", category_expense.idxmax())
print("Lowest Spending Category:", category_expense.idxmin())
print("Average Expense:", avg_expense)
```

## 🧠 Conclusion

- Monthly expense analysis helps understand spending behavior.
- Some categories contribute more to total expenses.
- Daily analysis shows which days have higher spending.
- Outliers indicate unusual expenses.
