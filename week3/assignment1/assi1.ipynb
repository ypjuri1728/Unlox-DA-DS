**1. What is a database in MySQL? Why is it important to organize data into
databases?**

* A database in MySQL is a structured collection of data stored in tables.

->Importance:

* Organizes data systematically
* Reduces data redundancy
* Makes data retrieval fast and easy
# * Improves data security

**2. Explain the difference between INT and BIGINT. When would you prefer one
over the other?**

ANS:
* INT: Stores smaller integer values (4 bytes)
* BIGINT: Stores very large integer values (8 bytes)

Use:

* INT → normal values (e.g., employee ID)
* BIGINT → very large values (e.g., large transactions, population)

**3. What is a PRIMARY KEY? Mention two rules that a primary key must follow.**
ans:

* A Primary Key is a column that uniquely identifies each row in a table.

Rules:

* Must be unique
* Cannot contain NULL values

**4. Differentiate between CHAR and VARCHAR with one practical example.**
ANS:

* CHAR: Fixed-length data type
* VARCHAR: Variable-length data type

Example:

->If storing "Hi"

* CHAR(10) → stores "Hi " (extra spaces)
* VARCHAR(10) → stores "Hi" only

**5. What is the purpose of NOT NULL and UNIQUE constraints?**

ANS:

* NOT NULL: Ensures that a column cannot have empty (NULL) values.
* UNIQUE: Ensures that all values in a column are different (no duplicates).

Purpose:

-->To maintain data accuracy and integrity in the table.
