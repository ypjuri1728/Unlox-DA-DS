**Answer the following questions:**

*1. What is the difference between DATE and DATETIME?*

Ans:
* DATE: Stores only date (YYYY-MM-DD)
* DATETIME: Stores date + time (YYYY-MM-DD HH:MM:SS)

Example:

* DATE → 2025-04-03
* DATETIME → 2025-04-03 21:30:00

*2. Why should email or phone number fields often be marked as UNIQUE?*

Ans:
* To avoid duplicate entries
* Ensures each user has one unique identity
* Maintains data accuracy and integrity

*3. Can a table have multiple UNIQUE constraints? Explain*

Ans:

-->Yes

* A table can have multiple UNIQUE columns
* Each column ensures no duplicate values

 Example: email and phone both can be UNIQUE
